# Getting Started with GX Core

We will learn how to:

- connect to data
- create expectations
- validate data
- review validation results

First of all, let's import the core library of Great Expectations.

```python
import great_expectations as gx
```

```python
context = gx.get_context()
```

```python
validator = context.sources.pandas_default.read_csv(
    "https://raw.githubusercontent.com/great-expectations/gx_tutorials/main/data/yellow_tripdata_sample_2019-01.csv"
)
```

```python
validator.head()
```

```python
validator.expect_column_values_to_be_between("passenger_count", min_value=1, max_value=6)
validator.expect_column_values_to_not_be_null("pickup_datetime")
validator.save_expectation_suite()
```

```python
checkpoint = context.add_or_update_checkpoint(
    name="my_quickstart_checkpoint",
    validator=validator,
)
```

```python
checkpoint_result = checkpoint.run()
```

```python
context.view_validation_result(checkpoint_result)
```

# Adding more expectations

Let's explore how we can go further by looking at https://greatexpectations.io/expectations/

E.g., we might expect that the values in the `payment_type` column are:
- all integers
- never null
- all belong to a predefined set of possible values

```python
validator.head()
```

```python
validator.expect_column_values_to_be_of_type("payment_type", "int")
validator.expect_column_values_to_not_be_null("payment_type")
validator.expect_column_values_to_be_in_set("payment_type", [1, 2, 3, 4])
```

```python
validator.save_expectation_suite(discard_failed_expectations=False)
checkpoint = context.add_or_update_checkpoint(
    name="my_customized_quickstart_checkpoint",
    validator=validator,
)
checkpoint_result = checkpoint.run()
```

```python
context.view_validation_result(checkpoint_result)
```
